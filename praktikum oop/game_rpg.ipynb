```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
    
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Anak (Hero)
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

# 3. Implementasi pada Class Anak (Monster)
class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# =========================================================================
# -- BLOK UJI COBA VALIDASI --
# =========================================================================
# unit = GameUnit() # ERROR! Abstract class tidak bisa diinstansiasi jadi objek.

h = Hero("Alucard")
m = Monster("Serigala")

print("--- Informasi Unit ---")
h.info()
m.info()

print("\n--- Simulasi Aksi ---")
h.serang("Lord")
m.serang("Alucard")
```

```python
# 1. INDUK CLASS (PARENT CLASS)
class Hero:
    def __init__(self, name):
        self.name = name

    def serang(self):
        print(f"{self.name} menyerang dengan tangan kosong.")


# 2. ANAK CLASS (CHILD CLASSES) - MENERAPKAN METHOD OVERRIDING
class Mage(Hero):
    def serang(self):
        print(f"🔥 {self.name} (Mage) menembakkan Bola Api! Boom!")


class Archer(Hero):
    def serang(self):
        print(f"🏹 {self.name} (Archer) memanah dari jauh! Jleb!")


class Fighter(Hero):
    def serang(self):
        print(f"⚔️ {self.name} (Fighter) memukul dengan pedang! Slash!")


# 3. PENERAPAN POLIMORFISME (POLYMORPHISM)
# Kita membuat daftar (list) pasukan yang berisi objek dari class yang berbeda
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")

# Hanya butuh SATU perintah loop, namun setiap objek merespons secara UNIK
for pahlawan in pasukan:
    pahlawan.serang()
```

```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name  # Nama Hero
        self.hp = hp  # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")
        
    # --- METHOD BARU UNTUK MENYERANG ---
    def serang(self, lawan):
        # Kurangi HP lawan dengan attack_power hero ini
        lawan.hp -= self.attack_power
        print(f"⚔️ {self.name} menyerang {lawan.name} sebesar {self.attack_power} DMG!")
        
        # Pastikan HP lawan tidak minus (mentok di 0 jika kalah)
        if lawan.hp < 0:
            lawan.hp = 0
            
        print(f"   👉 HP {lawan.name} sekarang tersisa: {lawan.hp}")


# -- Main Program --
# 1. Membuat Object (Instansiasi)
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# 2. Menampilkan Info Awal
print("--- Status Awal Hero ---")
hero1.info()
hero2.info()

# 3. Simulasi Pertarungan
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong (HP Zilong: 120 - 15 = 105)
hero2.serang(hero1) # Zilong membalas (HP Layla: 100 - 20 = 80)

print("\n--- Status Akhir Hero ---")
hero1.info()
hero2.info()
```

### 3. Eksperimen Fungsi super()

#### A. Analisis Error
* **Pesan Error:** `AttributeError: 'Mage' object has no attribute 'name'`
* **Penyebab:** Ketika baris `super().__init__(...)` dihapus, constructor milik class induk (`Hero`) menjadi tidak tereksekusi. Akibatnya, proses inisialisasi variabel dasar seperti `name`, `hp`, dan `attack_power` tidak pernah terjadi di memori untuk objek `Mage`, meskipun argumen "Eudora" sudah dikirimkan saat pembuatan objek. Class Mage hanya berhasil menginisialisasi atribut `mana` miliknya sendiri.

#### B. Kesimpulan Peran `super()`
Fungsi `super()` di dalam Python bertindak sebagai jembatan delegasi. Peran utamanya dalam pewarisan (inheritance) adalah untuk mengakses dan menjalankan metode atau constructor dari class induk (`superclass`) langsung dari dalam class anak (`subclass`). Hal ini memastikan seluruh atribut dan perilaku dasar dari induk diwarisi secara utuh tanpa perlu menulis ulang baris kode yang sama (*Don't Repeat Yourself*).

* **Analisis Makna Error:** Pesan tersebut mengindikasikan bahwa Python menolak proses pembuatan objek dari `class Hero`. Karena `Hero` menginduk pada `GameUnit` yang mewajibkan adanya metode `.serang()`, penghapusan metode tersebut dinilai sebagai pelanggaran kontrak pewarisan. Kelas anak tidak boleh diwujudkan selama blueprint abstraknya belum diimplementasikan secara konkret.
* **Konsekuensi Kelalaian:** Jika pengembang lupa menuliskan method yang telah dijanjikan oleh interface, program akan langsung mengalami kegagalan fatal (*runtime error*) saat inisialisasi objek dilakukan. Ini merupakan fitur proteksi OOP agar tidak ada objek cacat fungsi yang lolos ke dalam sistem game.

##  2: MENCETAK CETAKAN (`unit = GameUnit()`)

### 1. Analisis Larangan Instansiasi
* **Alasan Pelarangan:** `GameUnit` dikonstruksi sebagai kelas abstrak murni yang menggunakan dekorator `@abstractmethod`. Kelas ini hanya berfungsi sebagai konsep atau skema besar, bukan wujud nyata. Karena tidak memiliki implementasi logika internal pada fungsinya (hanya berisi pernyataan `pass`), pembuatan objek secara langsung diblokir oleh Python demi mencegah kebingungan eksekusi fungsi kosong.

### 2. Manfaat dan Kegunaan Kelas Abstrak
Walaupun tidak dapat memproduksi objek secara mandiri, `GameUnit` memiliki fungsi vital sebagai berikut:
1. **Arsitektur Pengendali (Kontrak Standar):** Menjamin keseragaman struktur dasar kode pada seluruh sub-class turunan (Hero, Monster, dll) agar memiliki nama metode dan parameter input yang identik.
2. **Pondasi Polimorfisme Bersekala Besar:** Memudahkan pengaturan koleksi objek yang beragam di dalam satu manajemen memori (misal: List unit game), sehingga eksekusi perintah massal dapat dijalankan secara dinamis tanpa perlu mengecek tipe spesifik masing-masing kelas.

### 1. Dampak Perubahan Nama Method
* **Gejala yang Timbul:** Karakter Archer (Miya) kehilangan kemampuan memanah spesifiknya dan justru mengeksekusi metode milik kelas induk, yaitu: `Miya menyerang dengan tangan kosong.`.

### 2. Esensi Keseragaman Nama pada Polimorfisme
Keseragaman nama method antar Parent dan Child Class merupakan syarat mutlak dalam polimorfisme karena nama tersebut adalah **kontrak komunikasi tunggal**. Sistem eksekusi massal (seperti perintah *looping*) dirancang untuk memicu satu nama metode yang seragam bagi seluruh objek. Jika keseragaman nama ini dilanggar, maka rantai polimorfisme objek akan terputus.
