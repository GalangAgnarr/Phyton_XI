```python
from abc import ABC, abstractmethod

# =========================================================================
# 1. ABSTRACTION & ENCAPSULATION (Parent Class / Abstract Class)
# =========================================================================
class BarangElektronik(ABC):
    def __init__(self, nama, harga_dasar, stok_awal):
        self.nama = nama
        # Encapsulation: Menggunakan Private Attribute (__)
        self.__harga_dasar = harga_dasar
        self.__stok = 0
        
        # Mengisi stok menggunakan method setter agar melewati validasi
        self.tambah_stok(stok_awal)

    # GETTER untuk membaca Harga Dasar (diperlukan oleh class anak)
    def get_harga_dasar(self):
        return self.__harga_dasar

    # GETTER untuk melihat stok
    def get_stok(self):
        return self.__stok

    # SETTER / METHOD untuk menambah stok dengan Validasi Keamanan
    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"❌ Gagal update stok {self.nama}! Stok tidak boleh negatif ({jumlah}).")
            return False
        else:
            self.__stok += jumlah
            return True

    # ABSTRACT METHODS (Wajib di-override oleh class anak)
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah):
        pass


# =========================================================================
# 2. INHERITANCE & POLYMORPHISM (Child Class: Laptop)
# =========================================================================
class Laptop(BarangElektronik):
    def __init__(self, nama, harga_dasar, stok_awal, processor):
        # Memanggil constructor induk lewat super()
        super().__init__(nama, harga_dasar, stok_awal)
        self.processor = processor

    # Overriding Method untuk detail spesifik Laptop
    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * 0.10
        print(f"[LAPTOP] {self.nama} | Proc: {self.processor}")
        print(f"   Harga Dasar: Rp {self.get_harga_dasar():,.0f} | Pajak(10%): Rp {pajak:,.0f}")

    # Overriding Method untuk hitung harga + pajak 10%
    def hitung_harga_total(self, jumlah):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 0.10
        return (harga_dasar + pajak) * jumlah


# =========================================================================
# 3. INHERITANCE & POLYMORPHISM (Child Class: Smartphone)
# =========================================================================
class Smartphone(BarangElektronik):
    def __init__(self, nama, harga_dasar, stok_awal, kamera):
        super().__init__(nama, harga_dasar, stok_awal)
        self.kamera = kamera

    # Overriding Method untuk detail spesifik Smartphone
    def tampilkan_detail(self):
        pajak = self.get_harga_dasar() * 0.05
        print(f"[SMARTPHONE] {self.nama} | Cam: {self.kamera}")
        print(f"   Harga Dasar: Rp {self.get_harga_dasar():,.0f} | Pajak(5%): Rp {pajak:,.0f}")

    # Overriding Method untuk hitung harga + pajak 5%
    def hitung_harga_total(self, jumlah):
        harga_dasar = self.get_harga_dasar()
        pajak = harga_dasar * 0.05
        return (harga_dasar + pajak) * jumlah


# =========================================================================
# 4. POLYMORPHISM FEATURE (Fungsi Luar untuk Keranjang Belanja)
# =========================================================================
def proses_transaksi(daftar_belanja):
    print("--- STRUK TRANSAKSI ---")
    total_tagihan = 0
    nomor = 1
    
    # Menerapkan Polimorfisme: Memproses list campuran dengan satu perintah seragam
    for item, jumlah_beli in daftar_belanja:
        print(f"{nomor}. ", end="")
        item.tampilkan_detail() # Polimorfisme aksi cetak detail
        
        subtotal = item.hitung_harga_total(jumlah_beli) # Polimorfisme hitung harga
        print(f"   Beli: {jumlah_beli} unit | Subtotal: Rp {subtotal:,.0f}")
        
        total_tagihan += subtotal
        nomor += 1
        
    print("-" * 40)
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")
    print("-" * 40)


# =========================================================================
# 5. SIMULASI ALUR PROGRAM (USER STORY)
# =========================================================================
print("--- SETUP DATA ---")
# 1) Admin membuat data produk awal
produk1 = Laptop("ROG Zephyrus", 20000000, 0, "Ryzen 9")
if produk1.tambah_stok(10):
    print(f"Berhasil menambahkan stok {produk1.nama}: {produk1.get_stok()} unit.")

produk2 = Smartphone("iPhone 13", 15000000, 0, "12MP")

# 2) Admin mencoba mengisi stok dengan angka negatif (Ditolak sistem)
produk2.tambah_stok(-5)

# Admin mengisi ulang dengan angka yang benar
if produk2.tambah_stok(20):
    print(f"Berhasil menambahkan stok {produk2.nama}: {produk2.get_stok()} unit.")

print()

# 3) User membeli barang (2 Laptop dan 1 Smartphone) dimasukkan ke list keranjang
# Format: (objek_barang, jumlah_beli)
keranjang_belanja = [
    (produk1, 2),
    (produk2, 1)
]

# 4) Panggil fungsi luar untuk mencetak struk transaksi akhir
proses_transaksi(keranjang_belanja)
```

---

## 2. Implementasi 4 Pilar OOP

### A. Abstraction (Abstraksi)
* Diimplementasikan melalui pembuatan `class BarangElektronik(ABC)`. Kelas induk ini bertindak sebagai kerangka murni (kontrak dasar) yang tidak bisa diubah langsung menjadi objek (`BarangElektronik()`). 
* Aturan standardisasi dipaksakan melalui dekorator `@abstractmethod` pada fungsi `tampilkan_detail()` dan `hitung_harga_total()`, yang menjamin seluruh jenis produk masa depan memiliki kedua fungsi krusial tersebut.

### B. Encapsulation (Enkapsulasi)
* Semua properti finansial dan inventaris diisolasi menggunakan tingkat akses privat via tanda *double underscore* (`self.__harga_dasar` dan `self.__stok`).
* Akses perubahan nilai diwajibkan melewati metode jembatan `tambah_stok()`. Metode ini dilengkapi dengan pengondisian logika (*validasi*) untuk mencegat input nilai negatif dari admin gudang guna mencegah malfungsi data inventaris.

### C. Inheritance (Pewarisan)
* Kelas spesifik `Laptop` dan `Smartphone` dibuat dengan mewarisi seluruh kapabilitas yang dimiliki oleh `BarangElektronik` melalui instruksi fungsi `super().__init__()`. Pewarisan ini memangkas redundansi penulisan kode atribut dasar nama, harga, dan stok.

### D. Polymorphism (Polimorfisme)
* **Method Overriding:** Kelas `Laptop` memodifikasi rumus penghitungan dengan menyertakan nilai pajak **10%**, sementara kelas `Smartphone` memodifikasinya menjadi sebesar **5%**.
* **Polymorphic Collection:** Fungsi eksternal `proses_transaksi()` memanfaatkan sifat polimorfisme di mana ia sanggup memproses sebuah struktur array list campuran berisi objek `Laptop` dan `Smartphone` secara bergantian menggunakan satu perintah perulangan tunggal (`item.tampilkan_detail()`), tanpa memerlukan seleksi tipe data manual (*if-else* jenis produk).
